```python
import numpy as np

class CarryPyramid:
    def __init__(self, layers=3):
        self.layers = layers

    def process_layer(self, data_tensor, delta_threshold):
        carry_data = []
        dropped_count = 0
        for data_point in data_tensor:
            density = np.linalg.norm(data_point)
            if density > delta_threshold:
                carry_data.append(data_point) # C: Carry
            else:
                dropped_count += 1            # D: Drop
        return np.array(carry_data), dropped_count

    def forward(self, input_data, base_delta):
        current_data = input_data
        total_dropped = 0

        for layer in range(self.layers):
            # 💡 층이 올라갈수록 임계치가 20%씩 가혹해짐
            layer_delta = base_delta * (1.0 + (layer * 0.2))

            print(f" [Layer {layer+1}] 심사 기준: {layer_delta:.2f} 이상만 통과 -> 심사 중...")

            current_data, dropped = self.process_layer(current_data, layer_delta)
            total_dropped += dropped

        return current_data, total_dropped

# ==========================================
# 🚀 시동 및 테스트 가동
# ==========================================
if __name__ == "__main__":
    pyramid = CarryPyramid(layers=3)

    # 1000개의 가상 데이터 (밀도는 평균적으로 1.0 ~ 1.2 부근에 집중됨)
    test_data = np.random.rand(1000, 3)

    print("=== Carry Pyramid 압축 엔진 가동 ===\n")

    # 방어 임계치(Delta)를 1.0으로 빡빡하게 올리고 가동!
    secured_data, dropped = pyramid.forward(test_data, base_delta=1.0)

    print("\n=== 최종 압축 결과 ===")
    print(f"들어온 원본 데이터 : {len(test_data)} 개")
    print(f"압축되어 살아남은 핵심 데이터(C) : {len(secured_data)} 개")
    print(f"물리적으로 소멸된 불필요한 데이터(D) : {dropped} 개")
```

```python
import numpy as np
import time
import random
import math
from IPython.display import clear_output

# =========================================================
# [2번 코드] Physical Ghost (코랩용 가상 온도 시뮬레이터 적용)
# =========================================================
class PhysicalGhostColab:
    def __init__(self, base_threshold=0.8):
        self.base_threshold = base_threshold
        self.simulated_temp = 30.0 # 초기 온도: 평온한 30도

    def read_cpu_temperature(self):
        # 핫팩을 댄 것처럼 1초마다 온도가 1.5 ~ 3.0도씩 비선형적으로 끓어오름
        self.simulated_temp += random.uniform(1.5, 3.0)
        if self.simulated_temp > 65.0:
            self.simulated_temp = 65.0 + random.uniform(-1.5, 1.5) # 65도에서 펄펄 끓며 진동
        return self.simulated_temp

    def generate_entropy_delta(self):
        current_temp = self.read_cpu_temperature()
        # 열역학적 노이즈를 방어 임계치에 주입
        thermal_noise = math.sin(current_temp * time.time()) * 0.15
        dynamic_delta = self.base_threshold + (current_temp * 0.015) + thermal_noise
        return dynamic_delta, current_temp

# =========================================================
# [3번 코드 통합 & 4번 데모 대시보드 UI]
# =========================================================
def simulate_pitch_demo(iterations=25):
    # 아까 메모리에 올라가 있는 CarryPyramid 엔진을 그대로 사용합니다.
    try:
        pyramid_engine = CarryPyramid(layers=3)
    except NameError:
        print("앗! 1번 클래스가 실행되지 않았습니다. 이전 셀을 먼저 실행해주세요.")
        return

    ghost_engine = PhysicalGhostColab(base_threshold=0.8)
    attack_count = 0

    try:
        for _ in range(iterations):
            attack_count += 1

            # 해커의 무차별 데이터 공격 (매번 1000개의 데이터 주입)
            incoming_data = np.random.rand(1000, 3)

            # 피지컬 고스트의 방어 임계치(온도) 변화
            dynamic_delta, current_temp = ghost_engine.generate_entropy_delta()

            # 캐리 피라미드 방어 수행
            secured_data, dropped = pyramid_engine.forward(incoming_data, base_delta=dynamic_delta)

            # 화면 클리어 및 UI 출력
            clear_output(wait=True)
            print("=" * 65)
            print(" 🛡️  NEXUS-NET: PHYSICAL GHOST SYSTEM ACTIVE  🛡️")
            print("=" * 65)

            # 온도가 오를수록 게이지가 붉은색(🔥)으로 타오름
            bar_length = int((current_temp - 30) / 1.5)
            temp_bar = "🔥" * bar_length if current_temp > 45 else "🟢" * bar_length

            print(f" [SYSTEM] Hardware Temp   : {current_temp:.1f} °C [{temp_bar}]")
            print(f" [SYSTEM] Defense Delta(Δ): {dynamic_delta:.4f} (Entropy injected)")
            print("-" * 65)
            print(f" [ATTACK] Hacker Tries    : {attack_count * 1000:,} vectors injected")
            print(f" [DEFENSE] Survived (C)   : {len(secured_data):,} nodes (Secured)")
            print(f" [DEFENSE] Burned (D)     : {dropped:,} nodes (Ash - Unrecoverable)")
            print("=" * 65)
            print(" 💡 관전 포인트: 온도가 오를수록 방어 기준(Delta)이 요동치며 데이터를 더 많이 태워버립니다.")

            time.sleep(1.0) # 1초 간격으로 스크린 갱신

        print("\n✅ 시연 종료: 물리적 노이즈 방어 시뮬레이션 성공")

    except KeyboardInterrupt:
        print("\n⏹️ 중지됨")

# 4번 데모 UI 가동! (25초 동안 시뮬레이션 됩니다)
simulate_pitch_demo(iterations=25)
```
