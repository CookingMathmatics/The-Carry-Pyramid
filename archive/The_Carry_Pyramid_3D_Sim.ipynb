```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 딥테크 논문에 어울리는 다크 모드 스타일 적용
plt.style.use('dark_background')

# ==========================================
# Fig 1. Blue Data (Raw Input) - 무작위 엔트로피 데이터
# ==========================================
np.random.seed(42)
# 50x50 크기의 0과 1로 이루어진 무작위 데이터 생성
raw_data = np.random.randint(0, 2, size=(50, 50))

plt.figure(figsize=(6, 6))
plt.imshow(raw_data, cmap='Blues', interpolation='nearest')
plt.title("Fig 1. Blue Data : Raw Input (Random Entropy)", fontsize=14, color='cyan')
plt.axis('off')
plt.tight_layout()
plt.show()

# ==========================================
# Fig 2. Modular Carry Operation (Z2 Filter) - 2D 단면
# ==========================================
size = 128
triangle = np.zeros((size, size))
triangle[0, size//2] = 1 # 피라미드 꼭대기 초기값

# Z2 공간에서의 파스칼 모듈로 연산 (Carry)
for i in range(1, size):
    for j in range(1, size-1):
        triangle[i, j] = (triangle[i-1, j-1] + triangle[i-1, j+1]) % 2

plt.figure(figsize=(8, 6))
plt.imshow(triangle, cmap='magma', interpolation='nearest')
plt.title("Fig 2. Z2 Filter : Modular Carry Dynamics", fontsize=14, color='orange')
plt.axis('off')
plt.tight_layout()
plt.show()

# ==========================================
# Fig 3. Physical Ghost (Fractal Emergence) - 3D 시어핀스키
# ==========================================
# 카오스 게임(Chaos Game) 알고리즘을 이용한 3D 프랙탈 생성
vertices = np.array([
    [0, 0, 0],
    [1, 0, 0],
    [0.5, np.sqrt(3)/2, 0],
    [0.5, np.sqrt(3)/6, np.sqrt(2/3)]
])

points = []
p = np.array([0.5, 0.2, 0.2]) # 임의의 시작점

# 15,000번의 연산을 통해 3D 토폴로지 구축
for i in range(15000):
    target = vertices[np.random.randint(0, 4)]
    p = (p + target) / 2
    points.append(p)

points = np.array(points)

fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
# 금빛 프랙탈 창발 효과
ax.scatter(points[:,0], points[:,1], points[:,2], s=0.5, c=points[:,2], cmap='Wistia', alpha=0.8)
ax.set_title("Fig 3. Physical Ghost : 3D Fractal Emergence", fontsize=14, color='gold')

# 깔끔한 출력을 위해 축 숨김
ax.set_xticks([])
ax.set_yticks([])
ax.set_zticks([])
ax.axis('off')
plt.tight_layout()
plt.show()
```
