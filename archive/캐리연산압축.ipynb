```python
import numpy as np

class CarryPyramid:
    def __init__(self, layers=3):
        self.layers = layers

    def process_layer(self, data_tensor, delta_threshold):
        carry_data = []
        dropped_count = 0
        for data_point in data_tensor:
            density = np.linalg.norm(data_point)
            if density > delta_threshold:
                carry_data.append(data_point) # C: Carry
            else:
                dropped_count += 1            # D: Drop
        return np.array(carry_data), dropped_count

    def forward(self, input_data, base_delta):
        current_data = input_data
        total_dropped = 0

        for layer in range(self.layers):
            # 💡 층이 올라갈수록 임계치가 20%씩 가혹해짐
            layer_delta = base_delta * (1.0 + (layer * 0.2))

            print(f" [Layer {layer+1}] 심사 기준: {layer_delta:.2f} 이상만 통과 -> 심사 중...")

            current_data, dropped = self.process_layer(current_data, layer_delta)
            total_dropped += dropped

        return current_data, total_dropped

# ==========================================
# 🚀 시동 및 테스트 가동
# ==========================================
if __name__ == "__main__":
    pyramid = CarryPyramid(layers=3)

    # 1000개의 가상 데이터 (밀도는 평균적으로 1.0 ~ 1.2 부근에 집중됨)
    test_data = np.random.rand(1000, 3)

    print("=== Carry Pyramid 압축 엔진 가동 ===\n")

    # 방어 임계치(Delta)를 1.0으로 빡빡하게 올리고 가동!
    secured_data, dropped = pyramid.forward(test_data, base_delta=1.0)

    print("\n=== 최종 압축 결과 ===")
    print(f"들어온 원본 데이터 : {len(test_data)} 개")
    print(f"압축되어 살아남은 핵심 데이터(C) : {len(secured_data)} 개")
    print(f"물리적으로 소멸된 불필요한 데이터(D) : {dropped} 개")
```
